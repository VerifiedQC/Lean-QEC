```python
def matrix_to_nat(matrix):
    # 1. Flatten the m x n matrix into a 1D array of m*n indices
    # 'C' order (row-major) is standard: [row0, row1, ...]
    flattened = np.array(matrix).flatten()
    
    # 2. Convert the bits to a binary string
    bit_string = ''.join(map(str, flattened))
    
    # 3. Convert binary string to a natural number (base 2)
    natural_number = int(bit_string, 2)
    
    return natural_number, bit_string
```

```python
m, n = 3, 3
binary_matrix = [
    [1, 0, 1],
    [0, 1, 1],
    [0, 0, 1]
]

val, bits = matrix_to_nat(binary_matrix)

print(f"Flattened Bits: {bits}")
print(f"Natural Number: {val}")
```

```python
def serialize_matrix(M):
  s = 0
  for k in range(M.shape[0] * M.shape[1]):
    i = k % M.shape[0]
    j = int(k / M.shape[0])
    if M[i, j]:
      s += (1 << k)
  return s
```
